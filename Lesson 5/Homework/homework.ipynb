### 2. The Tasks

Complete the following steps in a single Jupyter Notebook and submit it (with outputs visible).

#### Task 1: Load and Explore the Data
- Load the dataset.
- Print the shape of `X` and `y`.
- Print the feature names (if available).
- **Hint:** Check for missing values (if any, you may need to drop or impute them).

#### Task 2: Train a Baseline Traditional Model
- Split the data into training (80%) and testing (20%) using `train_test_split` with a fixed `random_state` (e.g., `42`).
- Train a **Linear Regression** (if regression) or **Logistic Regression / Decision Tree** (if classification) on the training set.
- Evaluate it on the **test** set using the appropriate metric:
  - **Regression:** Mean Squared Error (MSE) and R² Score.
  - **Classification:** Accuracy Score.
- Store this baseline performance—this is the number you must beat!

#### Task 3: Preprocess Data for the Neural Network
- **Feature Scaling (CRITICAL!):** Use `StandardScaler` to scale your features (`X`). Fit it **ONLY** on the training data and transform both train and test.
- **Target Scaling (for Regression only):** If doing regression, also scale your target variable `y` using `StandardScaler` (this helps the network converge smoothly). If doing classification, do NOT scale the labels.

#### Task 4: Build, Train, and Evaluate the Neural Network
- Convert the scaled NumPy arrays to PyTorch Tensors.
- Define a Neural Network class (`nn.Module`) with:
  - **At least 2 hidden layers** (e.g., `Linear(20)` → `ReLU` → `Linear(10)` → `ReLU` → `Output`).
  - Use **ReLU** as the activation function.
- Train the network for **at least 2000 epochs** using:
  - Loss function: `nn.MSELoss()` (regression) or `nn.CrossEntropyLoss()` (classification).
  - Optimizer: `torch.optim.SGD` or `torch.optim.Adam`.
  - **Learning Rate:** Experiment with at least 2 different learning rates (e.g., `0.01` and `0.001`) and observe the difference.
- During training, **store the loss at every epoch** to plot the "Mountain Descent" curve.

#### Task 5: Visualize the "Mountain Descent" (Loss Curve)
- Plot the training loss vs. epochs for BOTH learning rates on the same graph.
- Label them clearly (e.g., "LR = 0.01" and "LR = 0.001").
- **Reflection:** Which learning rate worked better? Did the loss bounce (Too Big) or crawl (Too Small)?

#### Task 6: Compare Performance
- Predict on the test set using your trained Neural Network.
- Inverse-transform the predictions back to the original scale (for regression) using `scaler_y.inverse_transform()`.
- Compute the same metric (MSE/R² or Accuracy) on the test set.
