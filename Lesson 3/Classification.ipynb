# Dataset

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)

df = iris.frame

df.head()
```

```python
X = df.drop(columns=["target"])
y = df["target"]
# target 0 is Setosa
# target 1 is Versicolor
# target 2 is Virginica
```

```python
X.head()
```

```python
y.head()
```

# Model

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()

model.fit(X, y)
```

```python
sample = [[5.1, 3.5, 1.4, 0.2]]

prediction = model.predict(sample)

print(prediction)
```

```python
predictions = model.predict(X)

print(predictions[:10].tolist())
print(y[:10].tolist())
```

```python
from sklearn.dummy import DummyClassifier

bad_model = DummyClassifier(strategy="most_frequent")
bad_model.fit(X, y)

bad_predictions = bad_model.predict(X)
```

```python
print(bad_predictions[:150].tolist())
print(y[:150].tolist())
```

```python
from sklearn.metrics import accuracy_score

print(
    accuracy_score(y, predictions)
)

print(
    accuracy_score(y, bad_predictions)
)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=iris.target_names,
    filled=True
)

plt.show()
```
