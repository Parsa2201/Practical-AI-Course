# Dataset

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)

df = housing.frame

df.head()
```

```python
X = df.drop(columns=["MedHouseVal"])
y = df["MedHouseVal"]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor()

model.fit(X, y)
```

```python
print(model.predict(X.iloc[:10]))

y.iloc[:10]
```

```python
y_pred = model.predict(X)
```

```python
import numpy as np
import matplotlib.pyplot as plt

dimension = 3
sample_range = slice(300, 350)

plt.figure(figsize=(6,4))
plt.scatter(X.to_numpy()[sample_range,dimension], y.to_numpy()[sample_range], c='red')
plt.plot(X.to_numpy()[sample_range,dimension], y_pred[sample_range])
plt.title("")
plt.show()
```
